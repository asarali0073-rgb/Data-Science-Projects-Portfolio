```python
import pandas as pd
import numpy as py
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('titanic_survival_classification_dataset.csv')
df
```

```python
df.info()
```

```python
df.describe()
```

# Null value

```python
df.isnull().sum()
```

# fillna

```python
df['Age'].fillna(df['Age'].mode()[0], inplace=True)
```

# Duplicat value

```python
df.duplicated().sum()
```

```python
df.dtypes
```

# outlier

```python
sns.set_theme()
sns.countplot(df,x='Sex')
plt.show()
```

```python
sns.pairplot(df)
```

```python
plt.figure(figsize=(8,6))
data = df.groupby('Sex')['Survived'].sum()
data.plot(kind='pie', autopct='%1.1f%%')
plt.title('Survived rating male and female')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.countplot(df,x ='Embarked')
plt.xticks(rotation=90)
plt.title('Embarked')
plt.show()
```

```python
plt.boxplot(df['Fare'])
plt.show()
```

```python
plt.figure(figsize=(8,6))
data = df.groupby('Sex')['Ticket'].sum()
data.plot(kind='pie', autopct='%1.1f%%')
plt.title('Survived rating male and female')
plt.tight_layout()
plt.show()
```
