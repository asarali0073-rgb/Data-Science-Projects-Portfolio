# Convert date and numeric columns

```python


# Convert date and numeric columns
data["Order Date"] = pd.to_datetime(
    data["Order Date"], errors="coerce"
)
data["Sales"] = pd.to_numeric(data["Sales"], errors="coerce")
data["Profit"] = pd.to_numeric(data["Profit"], errors="coerce")

data = data.dropna(subset=["Order Date", "Sales", "Profit"])
```


# BUSINESS PERFORMANCE SUMMARY


```python

total_revenue = data["Sales"].sum()
total_profit = data["Profit"].sum()
total_orders = data["Order ID"].nunique()
total_products = data["Product Name"].nunique()

profit_margin = (
    total_profit / total_revenue * 100
    if total_revenue != 0 else 0
)

print("========== BUSINESS PERFORMANCE REPORT ==========")
print(f"Total Revenue: ${total_revenue:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Total Orders: {total_orders:,}")
print(f"Unique Products: {total_products:,}")
print(f"Overall Profit Margin: {profit_margin:.2f}%")
```

```python
sns.set_theme(style="whitegrid")
```

# MONTHLY REVENUE TRENDS

```python

data["Month"] = data["Order Date"].dt.to_period("M")

monthly_sales = (
    data.groupby("Month", as_index=False)["Sales"]
    .sum()
)
monthly_sales["Month"] = monthly_sales["Month"].astype(str)

plt.figure(figsize=(12, 5))
sns.lineplot(
    data=monthly_sales,
    x="Month",
    y="Sales",
    marker="o"
)
plt.title("Monthly Revenue Trends")
plt.xlabel("Month")
plt.ylabel("Revenue ($)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("\n========== MONTHLY REVENUE ==========")
print(monthly_sales.to_string(index=False))

# Identify highest-revenue month
best_month = monthly_sales.loc[monthly_sales["Sales"].idxmax()]
print(
    f"\nHighest Revenue Month: {best_month['Month']} "
    f"(${best_month['Sales']:,.2f})"
)
```

# TOP 10 SELLING PRODUCTS BY REVENUE

```python

top_products = (
    data.groupby("Product Name", as_index=False)
    .agg(
        Revenue=("Sales", "sum"),
        Units_Sold=("Quantity", "sum"),
        Profit=("Profit", "sum")
    )
    .sort_values("Revenue", ascending=False)
    .head(10)
)

plt.figure(figsize=(12, 7))
sns.barplot(
    data=top_products,
    x="Revenue",
    y="Product Name"
)
plt.title("Top 10 Products by Revenue")
plt.xlabel("Total Revenue ($)")
plt.ylabel("Product")
plt.tight_layout()
plt.show()

print("\n========== TOP 10 PRODUCTS BY REVENUE ==========")
print(top_products.to_string(index=False))
```

# PROFIT ANALYSIS BY CATEGORY

```python

category_profit = (
    data.groupby("Category", as_index=False)
    .agg(
        Revenue=("Sales", "sum"),
        Profit=("Profit", "sum")
    )
    .sort_values("Profit", ascending=False)
)

plt.figure(figsize=(9, 5))
sns.barplot(
    data=category_profit,
    x="Category",
    y="Profit"
)
plt.title("Profit by Product Category")
plt.xlabel("Category")
plt.ylabel("Total Profit ($)")
plt.tight_layout()
plt.show()

print("\n========== PROFIT BY CATEGORY ==========")
print(category_profit.to_string(index=False))
```

# MOST PROFITABLE AND LOSS-MAKING PRODUCTS

```python


product_profit = (
    data.groupby("Product Name", as_index=False)
    .agg(
        Revenue=("Sales", "sum"),
        Profit=("Profit", "sum")
    )
)

print("\n========== TOP 10 MOST PROFITABLE PRODUCTS ==========")
print(
    product_profit.sort_values("Profit", ascending=False)
    .head(10)
    .to_string(index=False)
)

print("\n========== TOP 10 LOSS-MAKING PRODUCTS ==========")
print(
    product_profit.sort_values("Profit", ascending=True)
    .head(10)
    .to_string(index=False)
)
```

# BUSINESS INSIGHTS

```python

print("\n========== BUSINESS INSIGHTS ==========")

if total_profit >= 0:
    print("Overall business profit is positive.")
else:
    print("Overall business has a net loss.")

print(f"Overall profit margin is {profit_margin:.2f}%.")

best_product = product_profit.loc[product_profit["Revenue"].idxmax()]
print(
    f"Highest-revenue product: {best_product['Product Name']} "
    f"(${best_product['Revenue']:,.2f})"
)

best_profit_product = product_profit.loc[
    product_profit["Profit"].idxmax()
]
print(
    f"Most profitable product: {best_profit_product['Product Name']} "
    f"(${best_profit_product['Profit']:,.2f})"
)

loss_products = (product_profit["Profit"] < 0).sum()
print(f"Products with negative total profit: {loss_products}")
```
